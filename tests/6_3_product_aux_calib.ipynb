```python
import sys, os
repo_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
if repo_root not in sys.path:
    sys.path.insert(0, repo_root)
print("Added to sys.path:", repo_root)
from fixedincomelib import *
print("Fixed Income Library is loaded.")
value_date = "2026-03-26"
```

### Product Generic Forward

```python
print("Product creation and display:")
### creation
effective_date = "2026-03-26"
termination_date = "3M"
pay_or_rec = "receive"
currency = "USD"
notional = 1e6
index = "USD-LIBOR-BBA-3M-FLAT"
coupon = 0.05
accrual_basis = "Actual/360"
settlement_offset = "2D"
payment_offset = "2D"
business_day_convention = "MF"
holiday_convention = "NYC"
prod_generic_forward = qfCreateProductGenericForward(
    effective_date,
    termination_date,
    pay_or_rec,
    currency,
    notional,
    index,
    coupon,
    accrual_basis,
    settlement_offset,
    business_day_convention,
    holiday_convention,
    payment_offset,
    business_day_convention,
    holiday_convention,
    False)
display(qfDisplayProduct(prod_generic_forward))
print("Product (de-)serialization:")
### serialization
path = 'prod_generic_forward.pickle'
qfWriteProductToFile(prod_generic_forward, path)
# deserialize
prod_generic_forward_retrieved = qfReadProductFromFile(path)
# check
display(qfDisplayProduct(prod_generic_forward_retrieved))
# house keeping
os.remove(path)
### from data convention
print("Product From Data Convention:")
axis1 = "1Y"
values = 0.05
data_conv = "USD-LIBOR-BBA-3M-FLAT-FWD"
prod_generic_forward_from_conv = qfCreateProductFromDataConvention(
    value_date,
    data_conv,
    axis1,
    values
)
display(qfDisplayProduct(prod_generic_forward_from_conv))
path = 'prod_generic_forward_from_conv.pickle'
qfWriteProductToFile(prod_generic_forward_from_conv, path)
# deserialize
prod_generic_forward_from_conv_retrieved = qfReadProductFromFile(path)
# check
display(qfDisplayProduct(prod_generic_forward_from_conv_retrieved))
# house keeping
os.remove(path)
```
