```python
### make sure all display works properly
for conv_type in conv_type_list:
    these_convs = df_all_convs[df_all_convs.Type == conv_type].Name.unique().tolist()
    for this_conv in these_convs:    
        this_conv = DataConventionRegistry().get(this_conv)
        # check display issue
        try:
            this_conv.display()
        except:
            raise Exception(f"There is a display issue for {this_conv}.")
```
