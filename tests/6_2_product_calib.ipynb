### Product Cash Deposit

```python
print("Product creation and display:")
### creation
effective_date = "2026-03-26"
termination_date = "2026-05-26"
pay_or_rec = "receive"
currency = "USD"
notional = 1e6
coupon = 0.05
accrual_basis = "Actual/360"
payment_offset = "2D"
prod_cash_deposit = qfCreateProductCashDeposit(
    effective_date,
    termination_date,
    pay_or_rec,
    currency,
    notional,
    coupon,
    accrual_basis,
    payment_offset)
display(qfDisplayProduct(prod_cash_deposit))
print("Product (de-)serialization:")
### serialization
path = 'prod_cash_deposit.pickle'
qfWriteProductToFile(prod_cash_deposit, path)
# deserialize
prod_cash_deposit_retrieved = qfReadProductFromFile(path)
# check
display(qfDisplayProduct(prod_cash_deposit_retrieved))
# house keeping
os.remove(path)
### from data convention
print("Product From Data Convention:")
axis1 = "1Y"
values = 0.05
data_conv = "USD-CASH-DEPOSIT"
prod_cash_deposit_from_conv = qfCreateProductFromDataConvention(
    value_date,
    data_conv,
    axis1,
    values
)
display(qfDisplayProduct(prod_cash_deposit_from_conv))
path = 'prod_cash_deposit_from_conv.pickle'
qfWriteProductToFile(prod_cash_deposit_from_conv, path)
# deserialize
prod_cash_deposit_from_conv_retrieved = qfReadProductFromFile(path)
# check
display(qfDisplayProduct(prod_cash_deposit_from_conv_retrieved))
# house keeping
os.remove(path)
```
