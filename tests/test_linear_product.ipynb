**Note on `qfCreateProductFromDataConvention` for Zero Spread:** none of the 9 named `ZERO SPREAD` conventions in `data_conventions.json` currently resolve through `IndexRegistry` end-to-end -- 6 reference indices such as `SOFR-1B-CALIB` / `*-FLAT` that were never registered in `indices.json`, and 2 (`SOFR-1B-OVER-USD-ZERO-SPREAD`, `SONIA-1B-OVER-GBP-ZERO-SPREAD`) use a bare currency code (`USD`, `GBP`) as `reference_index`, which `ProductZeroSpread` always tries to resolve as an index. This is a pre-existing data gap, not something this notebook covers; the direct-construction test above remains the coverage for this product type.
