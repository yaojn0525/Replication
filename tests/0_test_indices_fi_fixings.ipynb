## Test Index Regitsry

```python
### index registry basic operations
display(qfListAllIndex())
qfClearIndexRegistry("USD-LIBOR-BBA-1M")
display(qfListAllIndex())
qfClearIndexRegistry()
qfReloadIndexRegistry()
display(qfListAllIndex())
```

```python
### register a specific index
index_name = "USD-TERM-SOFR-PROXY-1W"
content = {
    "type": "IBOR INDEX",
    "term": "1W",
    "convention": {
        "currency": "USD",
        "settlement offset": "2D",
        "settlement holidays": "USGS",
        "accrual basis": "Actual/360",
        "payment businessday convention": "F",
        "payment holiday convention": "USGS",
        "end of month": True
    }
}
qfRegisterIndex(index_name, content)
qfDisplayIndex(index_name)
```

```python
### pick one representative per index type
# ibor index
display(qfDisplayIndex("USD-LIBOR-BBA-3M"))
# overnight index
display(qfDisplayIndex("SOFR-1B"))
display(qfDisplayIndex("USD-Term SOFR Base-1B"))
# composite index
display(qfDisplayIndex("USD-SOFR-COMPOUND"))
# fx
display(qfDisplayIndex("GBP-USD"))
```

## Test FundingIdentifier Regitsry

```python
### funding registry basic operations
display(qfListAllFundingIdentifiers())
qfClearFundingIdentifierRegistry("JPY-GOVT-BOND-FUNDING")
display(qfListAllFundingIdentifiers())
qfClearFundingIdentifierRegistry()
qfReloadFundingIdentifierRegistry()
display(qfListAllFundingIdentifiers())
```

```python
### register a specific index
fi_name = "JPY-LIBOR-BBA-6M-FLAT"
content = {
    "currency": "JPY",
    "base index": "JPY-LIBOR-BBA-6M"
}
qfRegisterFundingIdentifier(fi_name, content)
qfDisplayFundingIdentifier(fi_name)
```

```python
### pick one representative per funding type
# currency funding
display(qfDisplayFundingIdentifier("USD"))
# index funding
display(qfDisplayFundingIdentifier("SOFR-1B-FLAT"))
### standalone funding
display(qfDisplayFundingIdentifier("USD-GOVT-BOND-FUNDING"))
```

### Fixings Management

```python
### list index fixings
index_name = 'usd-federal funds-h.15-1d'
qfListIndexFixings(index_name)
```

```python
### list index fixings for a specific start date and end date
index_name = 'usd-federal funds-h.15-1d'
qfReloadIndexFixings() # load all back for testing below
qfListIndexFixings(
    index_name,
    start_date='2025-06-17',
    end_date='2025-06-18'
)
```

```python
index_name = 'usd-federal funds-h.15-1d'
### insert index fixings
qfInsertIndexFixing(index_name, '2026-07-08', 0.02736)
display(qfListIndexFixings(index_name))
# insert more
dates = ['2026-07-09', '2026-07-10']
fixings = [0.02, 0.02]
qfInsertIndexFixing(index_name, dates, fixings)
display(qfListIndexFixings(index_name))
# remove index fixings
qfRemoveIndexFixings(index_name, dates)
display(qfListIndexFixings(index_name))
```

```python
### remove all fixings
# remove all index fixings
qfRemoveIndexFixings('*')
qfListAllIndexFixings()
```
