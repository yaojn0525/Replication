## Test Model Build Method

#### 1. YC IBOR Build Method

```python
# create yield curve index build method
build_method_type = "YC_IBOR_ELEMENT"
content = {
    "TARGET" : "USD-LIBOR-BBA-3M",
    "REFERENCE" : "",
    "FRA OR FIXING" : "USD-LIBOR-BBA-3M-FRA",
    "IBOR FUTURE" : "EURODOLLAR",
    "SWAP" : "USD-SWAP-SEMI-BOND"
}
yc_ibor_build_method = qfCreateBuildMethod(build_method_type, content)
display(qfDisplayModelBuildMethod(yc_ibor_build_method))
yc_ibor_build_method.swap
# serialize / de-serialize
path = 'yc_ibor_build_method.pickle'
qfWriteBuildMethodToFile(yc_ibor_build_method, path)
yc_ibor_build_method_ = qfReadBuildMethodFromFile(path)
# check
display(qfDisplayModelBuildMethod(yc_ibor_build_method_))
# house keeping
os.remove(path)
```

## Test Build Method Collection APIs

### qfCreateModelBuildMethodCollection

```python
bm_collection = qfCreateModelBuildMethodCollection(
    [yc_ibor_build_method, 
     yc_on_build_method, 
     yc_funding_build_method, 
     yc_fx_rate,
     yc_common])
assert bm_collection.num_build_methods == 5

retrieved_build_method = qfGetBuildMethodFromBuildMethodCollection(
    bm_collection, 'USD-LIBOR-BBA-3M', 'YC_IBOR_ELEMENT')
assert retrieved_build_method.type == "YC_IBOR_ELEMENT"\
      and retrieved_build_method.target == "USD-LIBOR-BBA-3M"
```

### qfWriteBuildMethodCollectionToFile / qfReadBuildMethodCollectionFromFile

```python
path = 'test_bm_collection.pickle'

result = qfWriteBuildMethodCollectionToFile(bm_collection, path)
assert result == 'DONE'
assert os.path.exists(path)

bm_col_back = qfReadBuildMethodCollectionFromFile(path)
assert bm_col_back.num_build_methods == bm_collection.num_build_methods

os.remove(path)
```
